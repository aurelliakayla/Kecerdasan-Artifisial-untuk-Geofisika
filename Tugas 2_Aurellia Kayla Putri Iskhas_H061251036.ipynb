### 1.1. Diketahui dan Ditanyakan

### 1.2. Dasar Teori
Pada konfigurasi Wenner, jarak antarelektroda berjarak sama ($a$). Beda potensial $\Delta V$ diukur di antara elektroda potensial $M$ dan $N$, sedangkan arus $I$ diinjeksikan melalui elektroda $A$ dan $B$.
* Rumus faktor geometri Wenner:
  $$K = 2\pi a$$
* Rumus resistivitas semu:
  $$\rho_a = K \cdot \frac{\Delta V}{I}$$
Nilai resistivitas semu mencerminkan sifat medium homogen ekivalen yang bergantung pada volume batuan yang dialiri arus listrik.

### 1.3. Pemeriksaan Data
* Seluruh satuan arus telah dinyatakan dalam Ampere ($\text{A}$) dan beda potensial dalam Volt ($\text{V}$).
* Rasio resistansi $\Delta V / I$ menurun seiring membesarnya spasi $a$, yang secara fisis mencerminkan pelebaran jalur arus pada volume medium yang lebih dalam. Data konsisten dan bebas dari anomali negatif.

### 1.4. Perhitungan Resistivitas Semu

* **Titik P4 ($a = 10.0\ \text{m}$):**

# 2. OUTPUT PROMPT 2 — Penentuan Kedalaman Lapisan dengan Seismik Refraksi

### 2.1. Diketahui dan Ditanyakan
* **Diketahui:**
  * Survei seismik refraksi 1 garis lintasan lurus, sumber di titik $x = 0\ \text{m}$.
  * Data jarak geophone ($x$) dan waktu tiba gelombang pertama ($t$):
    * G1: $x = 4.0\ \text{m}$, $t = 10.0\ \text{ms} = 0.0100\ \text{s}$
    * G2: $x = 8.0\ \text{m}$, $t = 20.0\ \text{ms} = 0.0200\ \text{s}$
    * G3: $x = 12.0\ \text{m}$, $t = 30.0\ \text{ms} = 0.0300\ \text{s}$
    * G4: $x = 16.0\ \text{m}$, $t = 35.0\ \text{ms} = 0.0350\ \text{s}$
    * G5: $x = 20.0\ \text{m}$, $t = 37.5\ \text{ms} = 0.0375\ \text{s}$
    * G6: $x = 24.0\ \text{m}$, $t = 40.0\ \text{ms} = 0.0400\ \text{s}$
* **Ditanyakan:**
  * Kecepatan lapisan 1 ($V_1$), kecepatan lapisan 2 ($V_2$), sudut kritis ($i_c$), waktu intersep ($t_i$), dan kedalaman batas lapisan ($h$).

### 2.2. Dasar Teori
* **Gelombang Langsung (*Direct Wave*):** Merambat di permukaan lapisan 1 dengan hubungan linier:
  $$t = \frac{x}{V_1}$$
* **Gelombang Refraksi Kritis (*Head Wave*):** Merambat di sepanjang bidang batas lapisan menurut hukum Snellius ($\sin i_c = V_1 / V_2$, syarat $V_2 > V_1$):
  $$t = \frac{x}{V_2} + t_i$$
* **Rumus Kedalaman Lapisan Intercept Time:**
  $$h = \frac{t_i V_1 V_2}{2\sqrt{V_2^2 - V_1^2}}$$

### 2.3. Pemeriksaan Data
Waktu tiba bertambah terhadap jarak geophone. Terlihat patahan gradien yang jelas pada $x = 12 - 16\ \text{m}$, yang mengonfirmasi bahwa geophone G1–G3 mencatat gelombang langsung, sedangkan G4–G6 mencatat gelombang refraksi kepala yang mendahului gelombang langsung (*first arrivals*).

### 2.4. Grafik Waktu Tempuh ($T$-$X$)
* **Segmen 1 (G1 – G3):** $\Delta t / \Delta x = (0.0300 - 0.0100) / (12 - 4) = 0.0200 / 8 = 0.0025\ \text{s/m}$.
* **Segmen 2 (G4 – G6):** $\Delta t / \Delta x = (0.0400 - 0.0350) / (24 - 16) = 0.0050 / 8 = 0.000625\ \text{s/m}$.
* *Crossover distance* berada di persimpangan kedua garis ($x_c \approx 13.3\ \text{m}$).

### 2.5. Perhitungan Kecepatan Lapisan
* **Kecepatan Lapisan 1 ($V_1$):**
  $$V_1 = \frac{1}{\text{Slope}_1} = \frac{1}{0.0025} = \mathbf{400.0\ \text{m/s}}$$
* **Kecepatan Lapisan 2 ($V_2$):**
  $$V_2 = \frac{1}{\text{Slope}_2} = \frac{1}{0.000625} = \mathbf{1600.0\ \text{m/s}}$$

### 2.6. Perhitungan Sudut Kritis dan Kedalaman
* **Sudut Kritis ($i_c$):**
  $$\sin i_c = \frac{V_1}{V_2} = \frac{400}{1600} = 0.250 \implies i_c = \arcsin(0.250) \approx \mathbf{14.48^\circ}$$
* **Waktu Intersep ($t_i$):**
  Menggunakan titik G6 ($x = 24.0\ \text{m}, t = 0.0400\ \text{s}$):
  $$t_i = t - \frac{x}{V_2} = 0.0400 - \frac{24.0}{1600.0} = 0.0400 - 0.0150 = \mathbf{0.0250\ \text{s}} = 25.0\ \text{ms}$$
* **Kedalaman Batas Antarmuka ($h$):**
  $$\sqrt{V_2^2 - V_1^2} = \sqrt{1600^2 - 400^2} = \sqrt{2560000 - 160000} = \sqrt{2400000} \approx 1549.19\ \text{m/s}$$
  $$h = \frac{0.0250 \times 400.0 \times 1600.0}{2 \times 1549.19} = \frac{16000}{3098.39} \approx \mathbf{5.16\ \text{meter}}$$

### 2.7. Tabel Hasil Perhitungan

| Lapisan | Kecepatan $V$ (m/s) | Sudut Kritis $i_c$ (°) | Waktu Intersep $t_i$ (s) | Kedalaman Batas $h$ (m) |
| :---: | :---: | :---: | :---: | :---: |
| **Lapisan 1** | 400.0 | — | — | $0 - 5.16\ \text{m}$ (Tebal: $5.16\ \text{m}$) |
| **Lapisan 2** | 1600.0 | $14.48^\circ$ | 0.0250 | $> 5.16\ \text{m}$ (*Bedrock/Half-space*) |

### 2.8. Interpretasi Geologi
* **Lapisan 1 ($V_1 = 400\ \text{m/s}$):** Lapisan tanah penutup tak terkonsolidasi (*loose weathered soil / alluvium*).
* **Lapisan 2 ($V_2 = 1600\ \text{m/s}$):** Batupasir lapuk sedang jenuh air atau lapisan batuan sedimen lempungan kompak yang bertindak sebagai batuan dasar geoteknik (*engineering bedrock*).

### 2.9. Evaluasi Ketidakpastian
Model mengasumsikan lapisan planar horizontal. Metode ini memiliki *blind-zone* alami dan tidak mampu mendeteksi lapisan kecepatan rendah di bawah lapisan kecepatan tinggi (*velocity inversion / hidden layer*).

### 2.10. Kesimpulan
Model dua lapisan menghasilkan batas kedalaman antarmuka lapisan lapuk pada kedalaman **$5.16\ \text{meter}$** dengan kontras kecepatan dari $400\ \text{m/s}$ menjadi $1600\ \text{m/s}$.

### 4.2. Dasar Teori
Analisis kerawanan longsor multikriteria menggabungkan parameter pengontrol kestabilan lereng menggunakan metode penjumlahan berbobot (*Weighted Linear Combination*):
$$S = \sum_{i=1}^n w_i \cdot r_i \quad \text{dengan } \sum w_i = 1$$
$S$ = Skor indeks kerawanan, $w_i$ = Bobot parameter ke-$i$, $r_i$ = Skor kelas parameter ke-$i$.

### 4.3. Pemeriksaan Data Spasial
Seluruh lapisan data di-proyeksikan ke sistem koordinat UTM Zona 50S (satuan meter), di-resample ke resolusi grid 30 meter, dan dibersihkan dari *NoData*.

### 4.4. Pengolahan Parameter
* Kemiringan lereng (*slope*) diturunkan dari DEM dalam satuan persen/derajat.
* Lapisan geologi dikelompokkan berdasarkan tingkat resistensi pelapukan batuan.

Ketidakpastian dipengaruhi oleh resolusi DEM 30m yang belum mencakup tebing buatan mikro di pinggir jalan raya.

### 4.10. Kesimpulan

### 5.6. Analisis Pola Anomali
Kurva anomali magnetik membentuk pola pasangan **dipol (kutub positif-negatif)** yang sangat tegas:
* Puncak positif maksimum berada di titik M02 ($+562.0\ \text{nT}$).
* Lembah negatif minimum berada di titik M04 ($-315.0\ \text{nT}$).
* Gradien anomali berubah sangat tajam antara M02 dan M03 ($\Delta(\Delta T) = 714\ \text{nT}$).

### 5.7. Interpretasi Geofisika
* Pola anomali dipolar bergradien curam antara M02 dan M03 mengindikasikan adanya tubuh intrusi batuan beku berkomposisi mafik/ultramafik (misalnya dike basal/andesit) yang kaya mineral magnetit.
* Perubahan tanda anomali yang drastis juga bertepatan dengan zona bidang patahan (*fault contact*) yang mendempetkan batuan beku bersuseptibilitas tinggi dengan batuan sedimen non-magnetik di sisi selatan.

### 5.8. Evaluasi Ketidakpastian
Bentuk pasangan anomali dipol sangat dipengaruhi oleh sudut inklinasi medan magnet bumi setempat. Selain suseptibilitas induksi, efek magnetisasi remanen alami (*Natural Remanent Magnetization* / NRM) dapat menggeser posisi anomali relatif terhadap tubuh batuan sumber sebenarnya.

### 5.9. Kesimpulan
Anomali magnetik total berkisar antara $-315.0$ hingga $+562.0\ \text{nT}$, menunjukkan indikasi kuat keberadaan struktur intrusi batuan beku atau bidang kontak patahan di antara stasiun M02 dan M03.
