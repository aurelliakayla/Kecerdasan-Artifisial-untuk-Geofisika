# CARA MENYUSUN PROMPT YANG BAIK

Prompt yang baik perlu disusun secara **jelas, terstruktur, spesifik, dan
memiliki tujuan yang terarah**. Prompt tidak hanya berisi perintah, tetapi
juga memberikan informasi mengenai peran yang harus dijalankan, tugas yang
harus dilakukan, konteks permasalahan, data yang digunakan, batasan dalam
proses penyelesaian, serta bentuk keluaran yang diharapkan.

Salah satu struktur yang dapat digunakan untuk menyusun prompt secara
sistematis adalah **R-T-C-I-K-O**, yaitu:

| Struktur | Kepanjangan | Fungsi |
|---|---|---|
| **R** | **Role** | Menentukan peran atau keahlian yang harus dijalankan AI |
| **T** | **Task** | Menjelaskan tugas atau tujuan yang harus diselesaikan |
| **C** | **Context** | Memberikan latar belakang dan konteks permasalahan |
| **I** | **Input** | Menentukan data atau informasi yang digunakan |
| **K** | **Constraints** | Menentukan batasan, aturan, dan ketentuan yang harus dipatuhi |
| **O** | **Output** | Menentukan bentuk dan struktur keluaran yang diharapkan |

Struktur R-T-C-I-K-O dapat digunakan untuk berbagai permasalahan, termasuk
permasalahan dalam bidang geofisika. Dengan struktur ini, AI dapat
memahami secara lebih jelas mengenai **peran yang harus dijalankan,
tugas yang harus dilakukan, konteks permasalahan, data yang tersedia,
batasan analisis, serta bentuk keluaran yang diinginkan**.

```text
R — Role
↓
T — Task
↓
C — Context
↓
I — Input
↓
K — Constraints
↓
O — Output


# PROMPT ANALISIS GEOFISIKA — PENENTUAN HIPOSENTER GEMPA

## R — ROLE

Anda berperan sebagai seorang **geofisikawan dan seismolog** yang memiliki
keahlian dalam analisis data gempa bumi, khususnya penentuan lokasi
hiposenter berdasarkan data waktu tiba gelombang seismik.

Anda harus melakukan analisis secara sistematis, ilmiah, transparan, dan
dapat ditelusuri. Setiap tahapan perhitungan harus dapat diverifikasi
berdasarkan data dan persamaan yang digunakan.

Anda harus membedakan dengan jelas antara:

- data observasi;
- parameter yang diberikan;
- asumsi;
- hasil perhitungan;
- interpretasi;
- ketidakpastian;
- keterbatasan metode.

Jangan mengarang data yang tidak diberikan.

---

# T — TASK

Tugas utama Anda adalah melakukan analisis untuk **menentukan hiposenter
gempa berdasarkan data waktu tiba gelombang P dan S pada beberapa stasiun
seismik**.

Lakukan tahapan berikut:

1. Identifikasi seluruh parameter yang diperlukan.
2. Periksa kelengkapan dan konsistensi data input.
3. Hitung selisih waktu tiba antara gelombang S dan P pada setiap stasiun.
4. Estimasi jarak sumber gempa terhadap masing-masing stasiun.
5. Gunakan informasi dari seluruh stasiun untuk memperkirakan posisi
   hiposenter.
6. Tentukan latitude, longitude, dan kedalaman hiposenter jika data
   memungkinkan.
7. Hitung residual antara data observasi dan hasil pemodelan apabila
   memungkinkan.
8. Evaluasi kualitas hasil penentuan hiposenter.
9. Jelaskan asumsi yang digunakan dalam analisis.
10. Jelaskan ketidakpastian dan keterbatasan hasil.
11. Tampilkan proses perhitungan secara bertahap.
12. Sajikan hasil akhir sesuai template Markdown pada bagian OUTPUT.

Jika data yang diberikan tidak cukup untuk menentukan hiposenter secara
unik, jangan membuat nilai hasil secara sembarangan. Jelaskan data apa
yang kurang dan sejauh mana analisis masih dapat dilakukan.

---

# C — CONTEXT

Permasalahan yang dianalisis adalah penentuan lokasi sumber gempa
berdasarkan waktu tiba gelombang seismik pada beberapa stasiun.

Gelombang P merupakan gelombang seismik yang umumnya tiba lebih dahulu,
sedangkan gelombang S tiba setelah gelombang P.

Selisih waktu tiba gelombang P dan S dapat digunakan untuk memperkirakan
jarak antara stasiun dengan sumber gempa.

Gunakan hubungan:

$$
\Delta t = t_S - t_P
$$

dengan:

- $\Delta t$ = selisih waktu tiba gelombang S dan P;
- $t_S$ = waktu tiba gelombang S;
- $t_P$ = waktu tiba gelombang P.

Jika kecepatan gelombang P dan S dianggap diketahui dan konstan, jarak
sumber terhadap stasiun dapat diperkirakan menggunakan:

$$
\Delta t =
\frac{r}{V_S} - \frac{r}{V_P}
$$

sehingga:

$$
r =
\frac{\Delta t}
{\frac{1}{V_S}-\frac{1}{V_P}}
$$

dengan:

- $r$ = jarak antara stasiun dan sumber gempa;
- $V_P$ = kecepatan gelombang P;
- $V_S$ = kecepatan gelombang S.

Gunakan persamaan tersebut hanya jika asumsi model kecepatan homogen
memang sesuai dengan data atau telah dinyatakan sebagai asumsi.

Jika tersedia model kecepatan yang lebih kompleks, gunakan pendekatan yang
sesuai dan jelaskan metode yang digunakan.

Dalam menentukan hiposenter, pertimbangkan koordinat stasiun, waktu tiba
gelombang, model kecepatan, serta kedalaman sumber.

---

# I — INPUT

Gunakan data yang diberikan oleh pengguna.

## Data Stasiun

| Stasiun | Latitude (°) | Longitude (°) | Elevasi (m) |
|---|---:|---:|---:|
| STA01 | [masukkan nilai] | [masukkan nilai] | [masukkan nilai] |
| STA02 | [masukkan nilai] | [masukkan nilai] | [masukkan nilai] |
| STA03 | [masukkan nilai] | [masukkan nilai] | [masukkan nilai] |
| STA04 | [masukkan nilai] | [masukkan nilai] | [masukkan nilai] |

## Data Waktu Tiba Gelombang

| Stasiun | $t_P$ (s) | $t_S$ (s) |
|---|---:|---:|
| STA01 | [masukkan nilai] | [masukkan nilai] |
| STA02 | [masukkan nilai] | [masukkan nilai] |
| STA03 | [masukkan nilai] | [masukkan nilai] |
| STA04 | [masukkan nilai] | [masukkan nilai] |

## Parameter Model

```text
Vp = [masukkan nilai] km/s
Vs = [masukkan nilai] km/s
Model kecepatan = [homogen/berlapis/lainnya]
Sistem koordinat = [WGS84/lainnya]


---

# K — CONSTRAINTS

Ikuti seluruh ketentuan berikut selama melakukan analisis:

1. Gunakan satuan yang konsisten untuk seluruh perhitungan.

2. Jika jarak dinyatakan dalam kilometer, gunakan kecepatan dalam km/s.

3. Jangan mengubah data input tanpa memberikan penjelasan.

4. Jangan mengarang nilai data, parameter, koordinat, waktu tiba,
   maupun hasil pengukuran.

5. Jika parameter diperlukan tetapi tidak tersedia, nyatakan secara
   eksplisit bahwa parameter tersebut tidak tersedia.

6. Jika diperlukan suatu asumsi, tuliskan asumsi tersebut secara jelas
   dan jelaskan alasan penggunaannya.

7. Tampilkan persamaan sebelum melakukan perhitungan.

8. Jelaskan setiap simbol yang terdapat dalam persamaan.

9. Tampilkan minimal satu contoh substitusi angka secara lengkap.

10. Pisahkan antara data observasi, asumsi, perhitungan, hasil, dan
    interpretasi.

11. Jika memungkinkan, hitung residual antara data observasi dan hasil
    pemodelan.

12. Jelaskan arti residual dan pengaruhnya terhadap kualitas hasil.

13. Jika terdapat ketidakpastian, tampilkan nilainya jika tersedia.
    Jika tidak tersedia, jelaskan sumber ketidakpastiannya secara
    kualitatif.

14. Jangan memberikan kesimpulan yang tidak didukung oleh data.

15. Jika jumlah atau distribusi stasiun tidak memadai untuk menentukan
    hiposenter secara unik, jelaskan keterbatasannya.

16. Jangan menganggap model bumi homogen sebagai fakta jika hal tersebut
    belum ditentukan. Jika digunakan, nyatakan sebagai asumsi.

17. Jika terdapat beberapa metode yang memungkinkan, jelaskan metode yang
    digunakan dan alasan pemilihannya.

18. Semua tabel harus menggunakan format Markdown.

19. Semua persamaan matematika harus menggunakan format LaTeX.

20. Jangan memberikan kode pemrograman kecuali diminta oleh pengguna.

21. Jangan langsung memberikan hasil akhir tanpa menunjukkan proses
    analisis.

22. Jangan mengganti data yang tidak tersedia dengan data buatan.

23. Jika hasil tidak dapat dihitung karena data kurang, tetap berikan
    struktur analisis dan tunjukkan bagian mana yang membutuhkan data
    tambahan.

24. Bedakan dengan jelas antara hasil numerik yang dihitung dari data
    dan interpretasi geofisika.


---

# O — OUTPUT

Sajikan hasil analisis dalam format Markdown berikut:

## 1. Dasar Teori

Tuliskan persamaan yang digunakan dan jelaskan setiap parameter.

## 2. Perhitungan

Tampilkan proses perhitungan secara sistematis dan berurutan.

Untuk setiap stasiun, tampilkan:

- waktu tiba gelombang P;
- waktu tiba gelombang S;
- selisih waktu P-S;
- perhitungan jarak sumber;
- hasil jarak sumber.

Gunakan format:

$$
\Delta t = t_S - t_P
$$

$$
r =
\frac{\Delta t}
{\frac{1}{V_S}-\frac{1}{V_P}}
$$

## 3. Penentuan Hiposenter

Tampilkan hasil dalam format:

| Parameter | Hasil | Satuan |
|---|---:|---|
| Latitude | [HASIL] | ° |
| Longitude | [HASIL] | ° |
| Kedalaman | [HASIL] | km |

## 4. Evaluasi Hasil

Jelaskan:

- residual hasil perhitungan;
- tingkat konsistensi hasil;
- ketidakpastian jika dapat dihitung;
- keterbatasan analisis.

## 5. Kesimpulan

Tuliskan kesimpulan berdasarkan hasil perhitungan tanpa menambahkan
informasi yang tidak berasal dari data.

---

# ATURAN OUTPUT

1. Gunakan format Markdown.
2. Gunakan LaTeX untuk persamaan matematika.
3. Tampilkan proses perhitungan sebelum hasil akhir.
4. Jangan mengarang nilai yang tidak tersedia.
5. Jika data tidak mencukupi, nyatakan data yang diperlukan.
6. Semua hasil numerik harus dapat ditelusuri dari data input.
